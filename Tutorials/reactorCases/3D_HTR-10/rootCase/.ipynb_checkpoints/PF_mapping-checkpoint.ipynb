```python
def read_fields(path):
    # Read cell volumes
    with open(path, 'r') as f:
        lines = f.readlines()

    # Find the starting and ending indices of the block
    start_idx = next(i for i, line in enumerate(lines) if '(' in line)
    end_idx = next(i for i, line in enumerate(lines[start_idx:]) if ';' in line) + start_idx - 1

    # Extract the block of lines
    block = lines[start_idx + 1:end_idx]

    # Use NumPy to efficiently process and store the data
    return np.array([float(line) for line in block])

def read_vectors(path):
    # Read cell volumes
    with open(path, 'r') as f:
        lines = f.readlines()

    # Find the starting and ending indices of the block

    start_idx = next(i for i, line in enumerate(lines) if '(' in line)
    end_idx = next(i for i, line in enumerate(lines[start_idx:]) if ';' in line) + start_idx - 1

    # Extract the block of lines
    block = lines[start_idx + 1:end_idx]

    # Use NumPy to efficiently process and store the data
    return np.array([[float(k) for k in line.split('(')[1].split(')')[0].split(' ')] for line in block])
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Define the x-axis tick labels as the variable names
data_peb['PF_p']=data_peb['PF']*100
x_axis = [1, 2, 3, 4, 5, 6]
x_labels_name = ['Radial position [m]', 'Elevation [m]', 'Pebble power [W]', 
                 'Packing fraction [%]', 'Relative variation\n in $T_{He}$ [%]', 
                 'Relative variation\n in $T_{fuel}$ [%]']
x_labels = ['r_dist', 'elev', 'peb_power', 'PF_p', 'var_T', 'var_Tf']

# Parameters
num_samples = 500  # Desired total number of samples
n_segments = 100   # Number of segments to ensure coverage of the range

# Normalize the data
data_fuel = data_peb.loc[(data_peb['peb_power'] > 0)&(data_peb['var_T']>2.4)]
data_fuel_normalized = (data_fuel - data_fuel.min()) / (data_fuel.max() - data_fuel.min())

# Special normalization for var_T and var_Tf with same scale
combined_min = min(data_fuel['var_T'].min(), data_fuel['var_Tf'].min())
combined_max = max(data_fuel['var_T'].max(), data_fuel['var_Tf'].max())

data_fuel_normalized['var_T'] = (data_fuel['var_T'] - combined_min) / (combined_max - combined_min)
data_fuel_normalized['var_Tf'] = (data_fuel['var_Tf'] - combined_min) / (combined_max - combined_min)

norm_var_Tf = (data_fuel['var_Tf'] - combined_min) / (combined_max - combined_min)

cmap = plt.get_cmap('coolwarm')
n_labels = 10

# Create segments for stratified sampling
bins = np.linspace(data_fuel['var_Tf'].min(), data_fuel['var_Tf'].max(), n_segments + 1)
digitized = np.digitize(data_fuel['var_Tf'], bins)  # Assign each sample to a segment

# Collect samples
samples = []
for i in range(1, n_segments + 1):
    segment_indices = np.where(digitized == i)[0]
    if len(segment_indices) > 0:
        segment_samples = np.random.choice(segment_indices, size=min(len(segment_indices), num_samples // n_segments), replace=False)
        samples.extend(segment_samples)

# Handle the case where there are fewer samples than desired

if len(samples) < num_samples:
    additional_samples = np.random.choice(np.setdiff1d(range(len(data_fuel)), samples), size=num_samples - len(samples), replace=False)
    samples.extend(additional_samples)

samples = np.unique(samples)

# Plot
plt.figure(figsize=(15, 6))
elevation = data_fuel['z'] - data_fuel['z'].min()  # Ensure minimum elevation is 0
for k in samples:
    color = cmap(norm_var_Tf.loc[data_fuel_normalized.index[k]])
    plt.plot(x_axis, 
             [
              data_fuel_normalized.loc[data_fuel_normalized.index[k], 'r_dist'],
              data_fuel_normalized.loc[data_fuel_normalized.index[k], 'elev'],
              data_fuel_normalized.loc[data_fuel_normalized.index[k], 'peb_power'], 
              data_fuel_normalized.loc[data_fuel_normalized.index[k], 'PF_p'], 
              data_fuel_normalized.loc[data_fuel_normalized.index[k], 'var_T'], 
              data_fuel_normalized.loc[data_fuel_normalized.index[k], 'var_Tf']],  
             color=color, alpha=0.75)

plt.xticks(x_axis, x_labels_name)
sc = plt.scatter([], [], c=[], cmap=cmap, vmin=data_fuel['var_Tf'].min(), vmax=data_fuel['var_Tf'].max())
plt.colorbar(sc, label='Relative variation\n in $T_{fuel}$ [%]')  # Link the colorbar to the correct mappable

for i, label in enumerate(x_labels):
    if label=='var_T' or label=='var_Tf':
        min_val = combined_min
        max_val = combined_max
        label_values = np.linspace(min_val, max_val, n_labels)
        plt.vlines(x_axis[i], 0, 1, color='gray', linestyle='dotted', alpha=0.3)
        for j, val in enumerate(label_values):
            y_position = j / (n_labels - 1)  # Normalize y positions between 0 and 1
            if i == 0:
                plt.text(x_axis[i] - 0.1, y_position, f'{val:.1f}', ha='right', va='center', fontsize=9)
            else:
                plt.text(x_axis[i] - 0.1, y_position, f'{val:.1f}', ha='right', va='center', fontsize=9)
            plt.hlines(y_position, x_axis[i] - 0.05, x_axis[i] + 0.05, color='gray', alpha=1)
    else:
        min_val = data_fuel[label].min()
        max_val = data_fuel[label].max()
        label_values = np.linspace(min_val, max_val, n_labels)
        plt.vlines(x_axis[i], 0, 1, color='gray', linestyle='dotted', alpha=0.3)
        for j, val in enumerate(label_values):
            y_position = j / (n_labels - 1)  # Normalize y positions between 0 and 1
            if i == 0:
                plt.text(x_axis[i] - 0.1, y_position, f'{val:.1f}', ha='right', va='center', fontsize=9)
            else:
                plt.text(x_axis[i] - 0.1, y_position, f'{val:.1f}', ha='right', va='center', fontsize=9)
            plt.hlines(y_position, x_axis[i] - 0.05, x_axis[i] + 0.05, color='gray', alpha=1)

plt.xlabel('Variables')
plt.ylabel('Normalized Values (0 to 1)')
plt.title('Normalized Dependency Plot with Color Gradient and Vertical Scales')
plt.gca().get_yaxis().set_visible(False)
plt.tight_layout()
plt.savefig('spider_normalized.png', dpi=400)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Define the x-axis tick labels as the variable names
x_axis = [1, 2, 3, 4, 5, 6]
x_labels_name = [ 'Radial position [m]', 'Elevation [m]', 'Pebble power [W]', 'Packing fraction [%]', 'Relative variation\n in $T_{He}$ [%]', 'Relative variation\n in $T_{fuel}$ [%]']
x_labels = ['r_dist', 'elev', 'peb_power', 'PF', 'var_T', 'var_Tf']

# Parameters
num_samples = 100  # Desired total number of samples
n_segments = 10  # Number of segments to ensure coverage of the range

# Normalize the data
data_fuel = data_peb.loc[(data_peb['peb_power'] > 0)&(data_peb['var_T']<2.4)]
data_fuel_normalized = (data_fuel - data_fuel.min()) / (data_fuel.max() - data_fuel.min())
norm_var_Tf = (data_fuel['var_Tf'] - data_fuel['var_Tf'].min()) / (data_fuel['var_Tf'].max() - data_fuel['var_Tf'].min())
cmap = plt.get_cmap('coolwarm')
n_labels = 10

# Create segments for stratified sampling
bins = np.linspace(data_fuel['var_Tf'].min(), data_fuel['var_Tf'].max(), n_segments + 1)
digitized = np.digitize(data_fuel['var_Tf'], bins)  # Assign each sample to a segment

# Collect samples
samples = []
for i in range(1, n_segments + 1):
    segment_indices = np.where(digitized == i)[0]
    if len(segment_indices) > 0:
        segment_samples = np.random.choice(segment_indices, size=min(len(segment_indices), num_samples // n_segments), replace=False)
        samples.extend(segment_samples)

# Handle the case where there are fewer samples than desired
if len(samples) < num_samples:
    additional_samples = np.random.choice(np.setdiff1d(range(len(data_fuel)), samples), size=num_samples - len(samples), replace=False)
    samples.extend(additional_samples)

samples = np.unique(samples)

# Create bins based on normalized var_Tf
data_fuel_normalized['bin'] = np.digitize(norm_var_Tf, bins)

# Calculate the mean and standard deviation for each bin
mean_trajectories = []
std_trajectories = []

n_samples_per_bin = 100# Choose how many trajectories to average within each bin
mean_trajectories = []
std_trajectories = []

# Calculate the mean and standard deviation for each bin
for i in range(1, n_segments + 1):
    bin_data = data_fuel_normalized[digitized == i]
    if not bin_data.empty:
        # Sample a subset from the bin to limit the number of trajectories
        sampled_bin_data = bin_data.sample(n=min(n_samples_per_bin, len(bin_data)), replace=False)
        mean_trajectory = sampled_bin_data[x_labels].mean()  # Mean for each variable in the bin
        std_trajectory = sampled_bin_data[x_labels].std()  # Standard deviation for each variable in the bin
        mean_trajectories.append(mean_trajectory)
        std_trajectories.append(std_trajectory)


# Plot
plt.figure(figsize=(15, 6))

# Plot the mean trajectories with standard deviation as a shaded region
for i, (mean_traj, std_traj) in enumerate(zip(mean_trajectories, std_trajectories)):
    color = cmap(i / (n_segments - 1))  # Color for each bin
    plt.plot(x_axis, mean_traj, color=color, alpha=0.5, label=f'Bin {i+1}')
    plt.fill_between(x_axis, mean_traj - std_traj, mean_traj + std_traj, color=color, alpha=0.2)  # Shaded area represents std

# Add x-axis labels and colorbar
plt.xticks(x_axis, x_labels_name)
sc = plt.scatter([], [], c=[], cmap=cmap, vmin=data_fuel['var_Tf'].min(), vmax=data_fuel['var_Tf'].max())
plt.colorbar(sc, label='Relative variation\n in $T_{fuel}$ [%]')  # Link the colorbar to the correct mappable

# Add vertical lines and bin labels for each variable
for i, label in enumerate(x_labels):
    min_val = data_fuel[label].min()
    max_val = data_fuel[label].max()
    label_values = np.linspace(min_val, max_val, n_labels)
    plt.vlines(x_axis[i], 0, 1, color='gray', linestyle='dotted', alpha=0.3)
    for j, val in enumerate(label_values):
        y_position = j / (n_labels - 1)  # Normalize y positions between 0 and 1
        if i == 0:
            plt.text(x_axis[i] - 0.1, y_position, f'{val:.1f}', ha='right', va='center', fontsize=9)
        else:
            plt.text(x_axis[i] - 0.1, y_position, f'{val:.1f}', ha='right', va='center', fontsize=9)
        plt.hlines(y_position, x_axis[i] - 0.05, x_axis[i] + 0.05, color='gray', alpha=1)

plt.xlabel('Variables')
plt.ylabel('Normalized Values (0 to 1)')
plt.title('Normalized Dependency Plot with Standard Deviation Shaded "Pipes"')
plt.gca().get_yaxis().set_visible(False)
plt.legend(title='Binned Averages', bbox_to_anchor=(1.05, 1), loc='upper left')  # Legend to distinguish bins
plt.tight_layout()
plt.savefig('pipe_plot.png', dpi=400)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import colors

# Define original r_id and z_id (existing matrix)

r_id=[4,5,6]
# z_id=[8,12,16,20,25,30]
z_id=z_id[::-1]
# New z_id (for extrapolation)
new_z_id =np.linspace(10,100,20)


# Linear extrapolation function
def extrapolate_data(matrix, original_z, new_z):
    extrapolated_matrix = []
    for row in matrix:
        # Interpolate existing data points, but allow extrapolation for points outside range
        extrapolated_row = np.interp(new_z, original_z[::-1], row[::-1], left=None, right=None)
        extrapolated_matrix.append(extrapolated_row)
    return np.array(extrapolated_matrix)

# Extrapolate all matrices to new z_id
T_av_extrapolated = extrapolate_data(T_av, z_id, new_z_id)
T_max_extrapolated = extrapolate_data(T_max, z_id, new_z_id)
Tf_av_extrapolated = extrapolate_data(Tf_av, z_id, new_z_id)
Tf_max_extrapolated = extrapolate_data(Tf_max, z_id, new_z_id)

# Updated function to plot the extrapolated matrix
def plot_heatmap(ax, data, title, z_id_labels, r_id_labels):
    ax.set_title(title)
    # Plot heatmap
    heatmap = ax.imshow(data, cmap=cmap, interpolation='nearest')

    # Create a smaller colorbar
    cbar = plt.colorbar(heatmap, ax=ax, orientation='horizontal', fraction=0.03, pad=0.15)
    cbar.ax.tick_params(labelsize=8)

    # Set axis labels for r_id and z_id
    ax.set_xticks(np.arange(len(z_id_labels)))
    ax.set_yticks(np.arange(len(r_id_labels)))
    ax.set_xticklabels([f"z_{int(z)-6}" for z in z_id_labels], rotation=90)
    ax.set_yticklabels([f"r_{r}" for r in r_id_labels])

# Create subplots for the 4 variables
fig, axs = plt.subplots(2, 2, figsize=(12, 8))

# Plot the extrapolated variables with the new z_id
cmap = plt.get_cmap("coolwarm")  # Colormap for visualization

plot_heatmap(axs[0, 0], T_av_extrapolated, "$\overline{T}_{He}$ [K]", new_z_id, r_id)
plot_heatmap(axs[0, 1], T_max_extrapolated, "$T^{max}_{He}$ [K]", new_z_id, r_id)
plot_heatmap(axs[1, 0], Tf_av_extrapolated, "$\overline{T}_{fuel}$ [K]", new_z_id, r_id)
plot_heatmap(axs[1, 1], Tf_max_extrapolated, "$T^{max}_{fuel}$ [K]", new_z_id, r_id)

# Adjust layout and show plot
plt.tight_layout()
plt.show()
```
